```python
# Define mappings
age_map = {1: '≤ 29 years', 2: '30-39 years', 3: '40-49 years', 4: '50-59 years', 5: '≥ 60 years'}
religion_map = {1: 'Islam', 2: 'Christian', 3: 'Buddha', 4: 'Hindu'}
ethnicity_map = {1: 'Malay', 2: 'Bumiputera', 3: 'Chinese', 4: 'Indians', 5: 'Other'}
marital_map = {1: 'First marriage', 2: '≥ Second marriage'}
marriage_period_map = {1: '≤ 5 years', 2: '6-10 years', 3: '≥ 11 years'}
children_map = {1: '0', 2: '1-2 children', 3: '≥ 3 children'}
education_map = {1: 'No schooling', 2: 'Incomplete primary', 3: 'Std 6', 4: 'Form 3', 5: 'Form 5',
                 6: 'Cert/Diploma', 7: 'Bachelors', 8: 'Masters', 9: 'Doctoral'}
employment_map = {1: 'Full-time', 2: 'Part-time', 3: 'Unemployed'}
working_map = {1: '< 1 year', 2: '1-3 years', 3: '3-5 years', 4: '5-10 years', 5: '> 10 years'}
income_map = {1: '< RM2,500', 2: 'RM2.5k-3.1k', 3: 'RM3.1k-3.9k', 4: 'RM3.9k-4.8k', 5: 'RM4.8k-5.8k',
              6: 'RM5.8k-7.0k', 7: 'RM7.1k-8.6k', 8: 'RM8.7k-10.9k', 9: 'RM10.9k-15.0k', 10: '≥ RM15k'}
```

```python
# Apply mappings
df['Age'] = df['Age'].map(age_map)
df['Religion'] = df['Religion'].map(religion_map)
df['Ethnicity'] = df['Ethnicity'].map(ethnicity_map)
df['Marital status'] = df['Marital status'].map(marital_map)
df['Marriage period'] = df['Marriage period'].map(marriage_period_map)
df['No. of children'] = df['No. of children'].map(children_map)
df['Educational level'] = df['Educational level'].map(education_map)
df['Employment status'] = df['Employment status'].map(employment_map)
df['Working period'] = df['Working period'].map(working_map)
df['Household income'] = df['Household income'].map(income_map)
```

```python
for dtype, cols in df.dtypes.groupby(df.dtypes):
    print(f"\n[{dtype}] 타입의 변수 목록:")
    print(list(cols.index))
```

# 명목형 - 인구통계코드

```python
socio_vars = ['Age', 'Religion', 'Ethnicity', 'Marital status', 'Marriage period', 'No. of children', 'Educational level', 'Employment status', 'Working period', 'Household income']

# bar + pie를 한 화면에 보이게 하는 함수 생성
def create_combined_charts(variable, data):
    value_counts = df[variable].value_counts()
    fig, axes = plt.subplots(1, 2, figsize=(14, 6))

    #bar chart
    sns.countplot(x=variable, data=df, palette='viridis', ax=axes[0])
    axes[0].set_xlabel(variable, fontsize=12)
    axes[0].set_ylabel('Count', fontsize=12)
    axes[0].set_title(f'{variable} - Bar Chart', fontsize=14)
    axes[0].tick_params(axis='x', rotation=45)

    # Pie Chart
    axes[1].pie(value_counts.values,
                labels=value_counts.index,
                autopct='%1.1f%%',
                startangle=90,
                textprops={'fontsize': 10})
    axes[1].set_title(f'{variable} - Pie Chart', fontsize=14)
    axes[1].axis('equal')  # 원형 유지

    plt.tight_layout()
    plt.show()

# 모든 변수에 대해 순서대로 bar + pie 함께 출력
for var in socio_vars:
    create_combined_charts(var, df)
```

```python
# 카테고리별 항목 그룹핑
stress_categories = {
    'Demand': ['OS1', 'OS2', 'OS3', 'OS4', 'OS5', 'OS6', 'OS7', 'OS8'],
    'Control': ['OS9', 'OS10', 'OS11', 'OS12', 'OS13', 'OS14'],
    'Manager Support': ['OS15', 'OS16', 'OS17', 'OS18', 'OS19'],
    'Peer Support': ['OS20', 'OS21', 'OS22', 'OS23'],
    'Role': ['OS24', 'OS25', 'OS26', 'OS27', 'OS28'],
    'Change': ['OS29', 'OS30', 'OS31'],
    'Relationships': ['OS32', 'OS33', 'OS34', 'OS35'],
    'Work-Family Conflict': ['OS36', 'OS37', 'OS38', 'OS39'],
    'Family-Work Conflict': ['OS40', 'OS41']
}

# 새로운 df 생성
stress_data = pd.DataFrame()

for category, columns in stress_categories.items():
    stress_data[category] = df[columns].mean(axis=1)

# box plot 시각화
plt.figure(figsize=(8, 4))
sns.boxplot(data=stress_data, palette="Set2")

plt.title('Box Plots for Occupational Stress Categories', fontsize=16)
plt.xlabel('Stress Categories', fontsize=14)
plt.ylabel('Scores', fontsize=14)
plt.xticks(rotation=45, fontsize=12)
plt.yticks(fontsize=12)
plt.tight_layout() 
plt.show()
```
